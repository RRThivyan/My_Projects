```python
from ultralytics import YOLO
```

```python
model = YOLO('yolov8m-cls.yaml').load('yolov8m-cls.pt')
```

```python
results = model.train(data='/content/drive/MyDrive/AI_Projects/Flower_Classification/images', epochs=50, imgsz=256)
```

```python
model = YOLO('/content/drive/MyDrive/AI_Projects/Flower_Classification/best.pt')
```

```python
import cv2
import os
import numpy as np
import matplotlib.pyplot as plt
```

```python
image_folder_path = '/content/drive/MyDrive/AI_Projects/Flower_Classification/test_images'

image_files = [f for f in os.listdir(image_folder_path) if os.path.isfile(os.path.join(image_folder_path, f))]
```

```python
 %matplotlib inline
for image_file in image_files:
  image_path = os.path.join(image_folder_path, image_file)
  predict = model.predict(image_path, save=True, save_txt=True)
  # cv2_imshow(predict[0].plot())
  img = cv2.cvtColor(predict[0].plot(), cv2.COLOR_BGR2RGB)
  plt.imshow(np.squeeze(img))
  plt.show()
```
